```python
from fileloder import read_multiple_pdfs
from langchain_text_splitters import RecursiveCharacterTextSplitter

DOC_PATHS = [
    "pdfs/PartA.pdf",
    "pdfs/PartB.pdf"
]

docs = read_multiple_pdfs(DOC_PATHS)

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)
splits = text_splitter.split_documents(docs)

print(f"  Number of pdfs : {len(docs)}")
print(f" Number of chunks after chenk :{len(splits)}")
print("\n---  fris Chunk ---")
print(splits[0].page_content)
```

```python
from dotenv import load_dotenv
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_core.vectorstores import InMemoryVectorStore

load_dotenv()

embeddings = GoogleGenerativeAIEmbeddings(model="models/gemini-embedding-001")
vectore_store = InMemoryVectorStore(embeddings)

# إضافة الأجزاء التي قسمناها في الخلية السابقة
document_ids = vectore_store.add_documents(documents=splits)
print(f"تم تخزين {len(document_ids)} جزء بنجاح في الـ Vector Store!")
```

```python
query = "three floors of one building ?"
results = vectore_store.similarity_search(query, k=2)

for i, res in enumerate(results, 1):
    print(f"=== Ruselt  {i} (resours : {res.metadata['source']}) ===")
    print(res.page_content[:400])
    print("\n")
```

```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from deepagents import create_deep_agent
from tool import search_documentation, backend  
load_dotenv()

# 1. تجهيز نموذج Gemini المجاني
llm = ChatGoogleGenerativeAI(model="Gemini 3.1 Flash TTS", temperature=0)

# 2. إنشاء الـ Deep Agent وربطه بالأداة والـ backend
agent = create_deep_agent(
    model=llm,
    tools=[search_documentation],
    backend=backend,
    system_prompt="You are a helpful assistant. Always use search_documentation to answer questions based on the PDFs."
)

# 3. تشغيل الـ Agent (هنا الـ LangGraph بيفتح الـ State فالأداة هتشتغل بدون الخطأ!)
response = agent.invoke({
    "messages": [{"role": "user", "content": "What are the main topics in the documents?"}]
})

# طباعة الإجابة النهائية
print(response["messages"][-1].content)
```
